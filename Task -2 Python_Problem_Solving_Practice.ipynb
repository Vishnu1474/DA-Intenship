**Q1. Find the Shortest Word in a Sentence**
1. Observe
Input: "Python makes coding easy"
Expected output: easy
2. Understand
Find the word with the minimum number of characters.
3. Break
- Split the sentence into words.
- Compare the lengths of the words.

**Q8. Find the Most Frequent Character**
1. Observe
Input: "banana"
Expected output: a
2. Understand
Find the character that appears most frequently, ignoring spaces. If frequencies tie, select the first character encountered.
3. Break
- Remove spaces.
- Count each character.
- Identify the character with the highest frequency.
- Preserve the first character in case of a tie.
4. Logic
- Use a dictionary to store character frequencies and a loop to find the most frequent character.

```python
text = "banana".replace(" ", "")
freq= {}
for char in text:
    freq[char] = freq.get(char, 0) + 1
most_freq = text[0]
for char in text:
    if freq[char] > freq[most_freq]:
        most_freq = char
print(most_freq)
```

```python
# Additional Test Case
text = "a b c a c".replace(" ", "")
frequency = {}
for char in text:
    frequency[char] = frequency.get(char, 0) + 1
most_frequent = text[0]
for char in text:
    if frequency[char] > frequency[most_frequent]:
        most_frequent = char
print(most_frequent)
```

**Q20. Find the Longest Word That Appears Only Once**
1. Observe
Input: "Python makes coding useful Python"
Expected output: coding
2. Understand
Find the longest word that occurs exactly once. Ignore case and return the first word if multiple qualifying words share the maximum length.
3. Break
- Convert the sentence to lowercase.
- Split the sentence into words.
- Find words that occur exactly once.
- Compare their lengths and select the longest.
4. Logic
- Use count() to identify unique words and len() to compare their lengths. Update the answer only when a strictly longer word is found.

```python
sentence = "Python makes coding useful Python"
words = sentence.lower().split()
longest = ""
for word in words:
    if words.count(word) == 1:
        if len(word) > len(longest):
            longest = word
print(longest)
```
